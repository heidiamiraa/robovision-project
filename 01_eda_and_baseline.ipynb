```python
from collections import Counter
import csv
import os

DATASET_DIR = r"C:\Users\USER\Desktop\mv folder\Challenge_Dataset"

ext_counts = Counter()
folder_counts = Counter()
total_files = 0
csv_files = []
json_files = []

for dirpath, dirnames, filenames in os.walk(DATASET_DIR):
  rel_dir = os.path.relpath(dirpath, DATASET_DIR)
  folder_counts[rel_dir] += len(filenames)
  for fname in filenames:
    total_files += 1
    ext = os.path.splitext(fname)[1].lower() or "<no_ext>"
    ext_counts[ext] += 1
    full_path = os.path.join(dirpath, fname)
    if ext == ".csv":
      csv_files.append(full_path)
    elif ext == ".json":
      json_files.append(full_path)

print(f"Total files: {total_files}")
print("\nTop folders & file counts:")
for folder, count in list(folder_counts.items())[:25]:
  print(f"  {folder}: {count}")

print("\nFile extensions:")
for ext, count in ext_counts.most_common():
  print(f"  {ext}: {count}")

for csv_path in csv_files[:5]:
  print(f"\n--- CSV Preview: {csv_path} ---")
  with open(csv_path, newline="", encoding="utf-8", errors="replace") as f:
    reader = list(csv.reader(f))
    print("Rows (incl. header if present):", len(reader))
    for row in reader[:6]:
      print(" ", row)

for json_path in json_files[:5]:
  print(
      f"\n--- JSON Path: {json_path} (size: {os.path.getsize(json_path)}"
      " bytes) ---"
  )
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

DATASET_DIR = r"C:\Users\USER\Desktop\mv folder\Challenge_Dataset"
train_df = pd.read_csv(os.path.join(DATASET_DIR, "train.csv"))
test_df = pd.read_csv(os.path.join(DATASET_DIR, "test.csv"))

train_sub, val_sub = train_test_split(
    train_df, test_size=0.2, stratify=train_df["task_type"], random_state=42
)
train_sub = train_sub.reset_index(drop=True)
val_sub = val_sub.reset_index(drop=True)

action_cols = [f"action_{i}" for i in range(6)]


def parse_targets(df):
  rows = []
  for _, r in df.iterrows():
    vec = []
    for col in action_cols:
      vec.extend([float(x.strip()) for x in str(r[col]).split(",")])
    rows.append(vec)
  return np.array(rows, dtype=np.float32)


y_train = parse_targets(train_sub)
y_val = parse_targets(val_sub)

print(
    "Train split:",
    len(train_sub),
    train_sub["task_type"].value_counts().to_dict(),
)
print(
    "Val split:  ", len(val_sub), val_sub["task_type"].value_counts().to_dict()
)
print(
    "y_train:",
    y_train.shape,
    "| y_val:",
    y_val.shape,
    "| NaNs:",
    np.isnan(y_train).sum(),
)
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torchvision import models, transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])


class ActionDataset(Dataset):

  def __init__(self, df, root_dir, targets=None, transform=None):
    self.df = df.reset_index(drop=True)
    self.root_dir = root_dir
    self.targets = targets
    self.transform = transform

  def __len__(self):
    return len(self.df)

  def __getitem__(self, idx):
    img_path = os.path.join(self.root_dir, self.df.loc[idx, "image_path"])
    with Image.open(img_path) as img:
      img = img.convert("RGB")
    if self.transform:
      img = self.transform(img)
    if self.targets is not None:
      return img, torch.tensor(self.targets[idx], dtype=torch.float32)
    return img, self.df.loc[idx, "image_id"]


train_ds = ActionDataset(train_sub, DATASET_DIR, y_train, transform)
val_ds = ActionDataset(val_sub, DATASET_DIR, y_val, transform)
test_ds = ActionDataset(test_df, DATASET_DIR, None, transform)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, num_workers=0)
val_loader = DataLoader(val_ds, batch_size=64, shuffle=False, num_workers=0)
test_loader = DataLoader(test_ds, batch_size=64, shuffle=False, num_workers=0)

model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
for param in model.parameters():
  param.requires_grad = False
model.fc = nn.Linear(model.fc.in_features, 24)
model = model.to(device)

criterion = nn.L1Loss()
optimizer = torch.optim.AdamW(model.fc.parameters(), lr=3e-3, weight_decay=1e-4)

epochs = 2
for epoch in range(1, epochs + 1):
  model.train()
  train_loss = 0.0
  for x, y in train_loader:
    x, y = x.to(device), y.to(device)
    optimizer.zero_grad()
    loss = criterion(model(x), y)
    loss.backward()
    optimizer.step()
    train_loss += loss.item() * len(x)

  model.eval()
  val_preds, val_targets = [], []
  with torch.no_grad():
    for x, y in val_loader:
      val_preds.append(model(x.to(device)).cpu().numpy())
      val_targets.append(y.numpy())

  val_preds = np.vstack(val_preds)
  val_targets = np.vstack(val_targets)
  val_mae = np.mean(np.abs(val_preds - val_targets))
  val_rmse = np.sqrt(np.mean((val_preds - val_targets) ** 2))
  print(
      f"Epoch {epoch}/{epochs} | Train MAE: {train_loss/len(train_ds):.5f} |"
      f" Val MAE: {val_mae:.5f} | Val RMSE: {val_rmse:.5f}"
  )

row_mae = np.mean(np.abs(val_preds - val_targets), axis=1)
sorted_idx = np.argsort(row_mae)

print("\nLowest-error validation samples (best):")
for idx in sorted_idx[:3]:
  print(
      f"  {val_sub.loc[idx, 'image_id']} ({val_sub.loc[idx, 'task_type']}) |"
      f" MAE: {row_mae[idx]:.5f}"
  )

print("\nHighest-error validation samples (worst):")
for idx in sorted_idx[-3:]:
  print(
      f"  {val_sub.loc[idx, 'image_id']} ({val_sub.loc[idx, 'task_type']}) |"
      f" MAE: {row_mae[idx]:.5f}"
  )

sample_sub = pd.read_csv(os.path.join(DATASET_DIR, "sample_submission.csv"))
test_pred_map = {}
model.eval()
with torch.no_grad():
  for x, img_ids in test_loader:
    preds = model(x.to(device)).cpu().numpy()
    for img_id, vec in zip(img_ids, preds):
      test_pred_map[img_id] = vec

sub_rows = []
for img_id in sample_sub["image_id"]:
  vec = test_pred_map[img_id]
  row_dict = {"image_id": img_id}
  for i in range(6):
    chunk = vec[i * 4 : (i + 1) * 4]
    row_dict[f"action_{i}"] = ",".join(f"{v:.4f}" for v in chunk)
  sub_rows.append(row_dict)

sub_out = pd.DataFrame(sub_rows)[sample_sub.columns]
sub_out.to_csv("submission.csv", index=False)
print("\nSaved submission.csv with shape:", sub_out.shape)
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from torchvision import models, transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])


class ActionDataset(Dataset):

  def __init__(self, df, root_dir, targets=None, transform=None):
    self.df = df.reset_index(drop=True)
    self.root_dir = root_dir
    self.targets = targets
    self.transform = transform

  def __len__(self):
    return len(self.df)

  def __getitem__(self, idx):
    img_path = os.path.join(self.root_dir, self.df.loc[idx, "image_path"])
    with Image.open(img_path) as img:
      img = img.convert("RGB")
    if self.transform:
      img = self.transform(img)
    if self.targets is not None:
      return img, torch.tensor(self.targets[idx], dtype=torch.float32)
    return img, self.df.loc[idx, "image_id"]


train_ds = ActionDataset(train_sub, DATASET_DIR, y_train, transform)
val_ds = ActionDataset(val_sub, DATASET_DIR, y_val, transform)
test_ds = ActionDataset(test_df, DATASET_DIR, None, transform)

train_loader = DataLoader(train_ds, batch_size=32, shuffle=True, num_workers=0)
val_loader = DataLoader(val_ds, batch_size=32, shuffle=False, num_workers=0)
test_loader = DataLoader(test_ds, batch_size=32, shuffle=False, num_workers=0)

model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
model.fc = nn.Linear(model.fc.in_features, 24)
model = model.to(device)

criterion = nn.L1Loss()
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)

epochs = 3
for epoch in range(1, epochs + 1):
  model.train()
  train_loss = 0.0
  for x, y in train_loader:
    x, y = x.to(device), y.to(device)
    optimizer.zero_grad()
    loss = criterion(model(x), y)
    loss.backward()
    optimizer.step()
    train_loss += loss.item() * len(x)

  model.eval()
  val_preds_list, val_targets_list = [], []
  with torch.no_grad():
    for x, y in val_loader:
      val_preds_list.append(model(x.to(device)).cpu().numpy())
      val_targets_list.append(y.numpy())

  val_preds = np.vstack(val_preds_list)
  val_targets = np.vstack(val_targets_list)
  val_mae = np.mean(np.abs(val_preds - val_targets))
  val_rmse = np.sqrt(np.mean((val_preds - val_targets) ** 2))
  print(
      f"Epoch {epoch}/{epochs} | Train MAE: {train_loss/len(train_ds):.5f} |"
      f" Val MAE: {val_mae:.5f} | Val RMSE: {val_rmse:.5f}"
  )

# Per-sample MAE for qualitative inspection (best vs. worst validation predictions)
row_mae = np.mean(np.abs(val_preds - val_targets), axis=1)
sorted_idx = np.argsort(row_mae)

print("\nLowest-error validation samples (best):")
for idx in sorted_idx[:3]:
  print(
      f"  {val_sub.loc[idx, 'image_id']} ({val_sub.loc[idx, 'task_type']}) |"
      f" MAE: {row_mae[idx]:.5f}"
  )

print("\nHighest-error validation samples (worst):")
for idx in sorted_idx[-3:]:
  print(
      f"  {val_sub.loc[idx, 'image_id']} ({val_sub.loc[idx, 'task_type']}) |"
      f" MAE: {row_mae[idx]:.5f}"
  )

# Save submission.csv matching sample_submission.csv
sample_sub = pd.read_csv(os.path.join(DATASET_DIR, "sample_submission.csv"))
test_pred_map = {}
model.eval()
with torch.no_grad():
  for x, img_ids in test_loader:
    preds = model(x.to(device)).cpu().numpy()
    for img_id, vec in zip(img_ids, preds):
      test_pred_map[img_id] = vec

sub_rows = []
for img_id in sample_sub["image_id"]:
  vec = test_pred_map[img_id]
  row_dict = {"image_id": img_id}
  for i in range(6):
    chunk = vec[i * 4 : (i + 1) * 4]
    row_dict[f"action_{i}"] = ",".join(f"{v:.4f}" for v in chunk)
  sub_rows.append(row_dict)

sub_out = pd.DataFrame(sub_rows)[sample_sub.columns]
sub_out.to_csv("submission.csv", index=False)
print("\nSaved submission.csv with shape:", sub_out.shape)
```
